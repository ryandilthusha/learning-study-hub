## 6.1 Debugging & Error Handling (try / except)

### Concept Summary
Debugging means finding and fixing mistakes in your code. Instead of letting your program crash when an error occurs, you can catch the error and handle it safely.

#### 1. Basic Debugging with `print()`
The fastest way to see why code fails is to print out your variables and check their values as the program runs.
- **Example:** `print("Current x:", x)` helps you see if `x` holds what you expect.

#### 2. The `try` Block
Put the code that might cause an error inside `try:`. Python will test running it.
- **Example:**
  ```python
  try:
      number = int("hello")  # This will fail!
  ```

#### 3. The `except` Block
Runs if an error happens inside `try:`. It catches the error so the program does not crash.
- **Catching specific errors:** It is best practice to name the error you expect (like `ValueError` or `ZeroDivisionError`).
- **Bare `except:`:** Catches all errors without naming one, but this is not recommended because it hides bugs.
- **Example:**
  ```python
  try:
      res = 10 / 0
  except ZeroDivisionError:
      print("You cannot divide by zero!")
  ```

#### 4. The `else` Block
Runs only if **no errors happened** in the `try:` block.
- **Example:**
  ```python
  try:
      val = 10 / 2
  except ZeroDivisionError:
      print("Error!")
  else:
      print("Success! Value is:", val)  # Output: Success! Value is: 5.0
  ```

#### 5. The `finally` Block
Always runs at the end, whether an error happened or not.
- **Example:**
  ```python
  try:
      x = 1
  finally:
      print("All done!")  # Always prints
  ```

````python
# --- Demo: Safe Division with try, except, else, finally ---

def safe_divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Error: Cannot divide by zero!")
    except TypeError:
        print("Error: Both inputs must be numbers!")
    else:
        print(f"Result of {a} / {b} is: {result}")
    finally:
        print("Calculation attempt finished.\n")

# 1. Successful run
safe_divide(10, 2)

# 2. Divide by zero run
safe_divide(10, 0)

# 3. Wrong type run
safe_divide(10, "two")
````

**Requirements:**
1. Write a function named `divide_100(user_text)`:
2. Inside a `try:` block:
   - Convert `user_text` into an integer.
   - Calculate `100 / number` and store it in `result`.
3. Add an `except ValueError:` block that prints `"Invalid input: not a number!"`.
4. Add an `except ZeroDivisionError:` block that prints `"Cannot divide by zero!"`.

````python
def divide_100(user_text):
    try:
        number = int(user_text)
        result = 100/number
    except ValueError:
        print("Input textt should be a number")
    except ZeroDivisionError:
        print("Can't divide by zero")
    else:
        print(f"Result: {result}")
    finally:
        print("Check Completed")



# Value Error check
divide_100("abc")

print("-"*20)

# Divide by zero check
divide_100("0")

print("-"*20)

# Actual Vallue
divide_100("25")
````

## 7.1 Reading and Writing Files (`with open` & File Modes)

### Concept Summary
Python uses `with open(...)` to handle files safely. It automatically closes the file when your code is done, preventing data loss.

#### 1. The `with open` Syntax
```python
file = open("filename.txt", "mode")
# Work with the file here
file.close()  # Don't forget to close the file!
```

#### 2. File Opening Modes
- **`"r"` (Read):** Opens a file for reading only. Raises an error if the file does not exist.
- **`"w"` (Write):** Opens a file for writing. If the file already exists, it **erases all old content** and starts fresh. If the file does not exist, it creates it.
- **`"a"` (Append):** Adds new text to the end of the file. It **preserves existing content** without erasing it.
- **`"r+"` (Read & Write):** Allows both reading and writing. Keeps existing content (does not truncate).
- **`"w+"` (Write & Read):** Allows both writing and reading, but it **erases all old content** immediately upon opening!
- **`"b"` (Binary):** Used for non-text files like images or audio.

````python
# --- Demo: Writing, Appending, and Reading with Modes ---


# 1. 'w' mode: Writes initial content (creates file)
file = open("./supporting_files/demo.txt", "w")
file.write("Line 1: Hello\n")
file.close()

# 2. 'a' mode: Appends new content without deleting Line 1
file = open("./supporting_files/demo.txt", "a")
file.write("Line 2: World\n")
file.close()

# 3. 'r' mode: Reads back the full content
file = open("./supporting_files/demo.txt", "r")
content = file.read()
print("File content:\n" + content)
file.close()
````

## 7.2 File Reading & Writing Methods

### Concept Summary
Python gives you different methods to read and write data depending on whether you want a single string or a list of lines.

#### 1. Writing Text: `.write()` vs. `.writelines()`
- **`file.write(text)`**: Writes a single string into the file.
  - **Example:** `f.write("Hello\n")`
- **`file.writelines(list_of_strings)`**: Writes a list of strings into the file all at once.
  - **Example:** `f.writelines(["Line 1\n", "Line 2\n"])`

#### 2. Reading Text: `.read()`, `.readline()`, and `.readlines()`
- **`file.read()`**: Reads the **entire file** and returns it as one single string.
  - **Example:** `content = f.read()`
  - **Output:** `"Line 1\nLine 2\n"`

- **`file.readline()`**: Reads just **one line** at a time as a string. Calling it again reads the next line.
  - **Example:** `first_line = f.readline()`
  - **Output:** `"Line 1\n"`
- **`file.readlines()`**: Reads **all lines** and returns them as a **list of strings**.
  - **Example:** `lines = f.readlines()` $\rightarrow$ `['Line 1\n', 'Line 2\n']`

````python
# --- Demo: Comparing File Methods ---


# 1. writelines(): Write a list of lines
fruits = ["Apple\n", "Banana\n", "Cherry\n"]
file = open("./supporting_files/demo_fruits.txt", "w")
file.writelines(fruits)
file.close()


# 2. readline(): Read just the first line
file = open("./supporting_files/demo_fruits.txt", "r")
first_line = file.readline()
print("First line:  ", first_line.strip())          # Output: "Apple"
second_line = file.readline()
print("Second line: ", second_line.strip())         # Output: "Banana"
file.close()


# 3. readlines(): Read all lines into a list
file = open("./supporting_files/demo_fruits.txt", "r")
all_lines = file.readlines()
print("All lines:", all_lines)  # Output: ['Apple\n', 'Banana\n', 'Cherry\n']
file.close() 
````
