### Exercise 9.1: Find the Matching List Comprehension

**Question:**
Which of the following lines of code produces the list: `[2, 6, 12, 20, 30, 42]`?

## 9.2 Skipping Errors

### Exercise 9.2: Identify Error-Handling Syntax

**Question:**
Which of the following blocks catches an exception, prevents a crash, and allows code execution to continue?

- **A)** `while:`
- **B)** `try: except:`
- **C)** `for:`

<details>
<summary>💡 Need a hint?</summary>

- Two of these are loops used for repeating tasks.
- Only one construct is designed specifically for exception handling.
</details>

````python
# LEt's try to divide 10 with 0

# A) while loop
x = 10
y = 2
while x > -2:
    result = x/y
    print(result)
    x-=1
    y-=1
````

````python
# c) for loop

y = 2
for x in range(10,1,-1):
    result = x/y
    print(result)
    y-=1
````

````python
# B) try except

try:
    y = 2
    for x in range(10,1,-1):
        result = x/y
        print(result)
        y-=1
except ZeroDivisionError:
    print("Can't divide by 0")
````

## 9.3 Janky Functions

### Exercise 9.3: Default Mutable Arguments

**Question:**
What will be printed by the following code snippet?

```python
def add_to_list(item, lst=[]):
    lst.append(item)
    return lst

list1 = add_to_list(1)
list2 = add_to_list(2, [])
list3 = add_to_list(3)
print(list1, list2, list3)
```

- **A)** `[1, 3] [2] [1, 3]`
- **B)** `[1] [2] [1, 3]`
- **C)** `[1] [2] [3]`
- **D)** `[1] [1, 2] [1, 2, 3]`

<details>
<summary>💡 Need a hint?</summary>

- When is the default parameter `lst=[]` created: every function call, or once when defined?
- Notice which calls omit the second argument (using the default) and which call provides a fresh empty list `[]`.
- `list1` and `list3` reference the exact same underlying object in memory.
</details>

````python
def add_to_list(item, lst=[]):
    lst.append(item)
    return lst

list1 = add_to_list(1)
list2 = add_to_list(2, [])      # Pass an empty list to avoid using the default list. 
list3 = add_to_list(3)
print(list1, list2, list3)
````

## 9.5 Is it Prime?

### Exercise 9.5: Prime Number Checker

**Goal:** Determine whether a given number is a prime number.

**Requirements:**
1. Prompt the user with: `"Give a number: "` and convert it to an integer.
2. A prime number is an integer greater than 1 that is divisible only by 1 and itself.
3. If prime, display: `"<number> is a prime number"`.
4. If not prime, display: `"<number> is not a prime number"`.

**Example Run:**
```text
Give a number: 7
7 is a prime number
```

<details>
<summary>💡 Need a hint?</summary>

- Numbers less than or equal to 1 are never prime.
- Try checking possible divisors from 2 up to `number - 1`.
- Use the modulo operator `%` to check if a number divides evenly with zero remainder.
</details>

````python
number_str = input("Give a number: ")
number = int(number_str)

if number>1:
    if number%2==0:
        print(f"{number} is not a prime number")
    else:
        print(f"{number} is a prime number")
elif number ==1:
    print(f"{number} is not a prime number")
````

## 9.6 Bank Account

### Exercise 9.6: Bank_account Class

**Goal:** Build a class that models an account, tracks balance, and prevents overdrafts.

**Requirements:**
1. Create a class called `Bank_account`.
2. In `__init__`, set up instance attributes for `name` and `balance`.
3. Implement an `add(amount)` method that adds money and prints confirmation.
4. Implement a `spend(amount)` method that checks if sufficient balance exists:
   - If yes: deduct the amount and print confirmation.
   - If no: print `"Insufficient funds"` without modifying the balance.
5. Implement `__str__` returning account summary details.

**Example Output Structure:**
```text
Name: Andrew Andrewson
Balance: 100
25
Added 25 euros to the account
2
Added 2 euros to the account
170
Insufficient funds
90
Spent 90 euros
Account name: Andrew Andrewson, Account balance: 37
```

<details>
<summary>💡 Need a hint?</summary>

- In `spend()`, use an `if / else` check to compare the requested amount against `self.balance`.
- `__str__` must return a string (not print it directly).
</details>

````python
class Bank_account:

    def __init__(self, name, balance):
        self.name = name
        self.balance = balance

    def add(self, amount):
        print(amount)
        self.balance += amount
        print(f"Added {amount} euros to the account")

    def spend(self, amount):
        print(amount)
        if amount<self.balance:
            self.balance-=amount
            print(f"Spent {amount} euros")
        elif amount>=self.balance:
            print("Insufficient funds")

    def __str__(self):
        return f"Account name: {self.name}, Account balance: {self.balance}"


cus1 = Bank_account("Andrew Andrewson", 100)
print("Name:", cus1.name)
print("Balance:", cus1.balance)

cus1.add(25)
cus1.add(2)
#print(cus1)
cus1.spend(170)
cus1.spend(90)
print(cus1)
    
````
