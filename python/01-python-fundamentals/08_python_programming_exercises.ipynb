## 9.1 Creating Lists

### Concept Summary
A list comprehension builds a new list by running a math formula for every number in a range.
- `range(start, stop)` starts at `start` and stops right before `stop`.
- **Example:** `range(1, 4)` gives `1, 2, 3`.
- **Example:** `[x + 1 for x in range(1, 4)]` $\rightarrow$ `[2, 3, 4]`

<div style="height: 20px; background: linear-gradient(to bottom, red 0%, red 2px, rgba(255, 0, 0, 0.5) 3px, transparent 100%); border-radius: 6px 6px 0 0; margin: 15px 0;"></div>

### Exercise 9.1: Find the Matching List Comprehension

**Question:**
Which of the following lines of code results in the list: `[2, 6, 12, 20, 30, 42]`?

- A) `[num*2 for num in range(6)]`
- B) `[num**2 + 2 for num in range(1,6)]`
- C) `[num**2 + num for num in range(6)]`
- D) `[num**2 + num for num in range(1,7)]`

<details>
<summary>💡 Need a hint?</summary>

- Option D runs `num` from 1 to 6:
  - For 1: $1^2 + 1 = 2$
  - For 2: $2^2 + 2 = 6$
  - For 3: $3^2 + 3 = 12$
  - For 4: $4^2 + 4 = 20$
  - For 5: $5^2 + 5 = 30$
  - For 6: $6^2 + 6 = 42$
</details>

## 9.2 Skipping Errors

### Concept Summary
When Python runs into an error, it stops and crashes unless you handle it.
- `try:` lets you test a piece of code.
- `except:` catches the error if one occurs and lets the program keep running without crashing.

## 9.3 Janky Functions

### Concept Summary
Default arguments like `lst=[]` are created **only once** when Python defines the function.
- Any call that relies on the default list modifies that exact same list in memory.
- Passing a fresh list like `add_to_list(2, [])` does not touch the default list.

<div style="height: 20px; background: linear-gradient(to bottom, red 0%, red 2px, rgba(255, 0, 0, 0.5) 3px, transparent 100%); border-radius: 6px 6px 0 0; margin: 15px 0;"></div>

### Exercise 9.3: Output of Default Mutable Arguments

**Question:**
What is the output of the following code?
```python
def add_to_list(item, lst=[]):
    lst.append(item)
    return lst

list1 = add_to_list(1)
list2 = add_to_list(2, [])
list3 = add_to_list(3)
print(list1, list2, list3)
```

- A) `[1, 3] [2] [1, 3]`
- B) `[1] [2] [1, 3]`
- C) `[1] [2] [3]`
- D) `[1] [1, 2] [1, 2, 3]`

<details>
<summary>💡 Need a hint?</summary>

- `list1` uses the default list `lst`, adding `1` -> `[1]`.
- `list2` uses a new empty list `[]`, adding `2` -> `[2]`.
- `list3` uses the default list again, adding `3` -> `[1, 3]`.
- Because `list1` and `list3` point to the same default list, both now show `[1, 3]`.
</details>

## 9.5 Is it Prime?

### Concept Summary
A prime number is a whole number greater than 1 that cannot be divided evenly by any number except 1 and itself.
- `7` is prime (only $1 \times 7$).
- `4` is not prime (divisible by 2: $4 \% 2 == 0$).

<div style="height: 20px; background: linear-gradient(to bottom, red 0%, red 2px, rgba(255, 0, 0, 0.5) 3px, transparent 100%); border-radius: 6px 6px 0 0; margin: 15px 0;"></div>

## 9.7 Fibonacci with Recursion

### Concept Summary
The Fibonacci sequence starts `0, 1, 1, 2, 3, 5, 8...` where each step is the sum of the previous two.
- **Base Cases:**
  - Step 0 $\rightarrow$ `0`
  - Step 1 $\rightarrow$ `1`
- **Recursive Case:** `fibonacci(t - 1) + fibonacci(t - 2)`

<div style="height: 20px; background: linear-gradient(to bottom, red 0%, red 2px, rgba(255, 0, 0, 0.5) 3px, transparent 100%); border-radius: 6px 6px 0 0; margin: 15px 0;"></div>

2. Prompt the user: `t = int(input("What step should we calculate? "))`.

<details>
<summary>💡 Need a hint?</summary>
