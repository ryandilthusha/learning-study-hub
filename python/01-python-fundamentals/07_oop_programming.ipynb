## 8.1 Classes, Attributes, and Methods

### Concept Summary
Object-Oriented Programming (OOP) groups related data and functions together into reusable blueprints called **Classes**.

#### 1. What is a Class and an Object?
- **Class:** The master blueprint (e.g., the architectural plan for a house).
- **Object (Instance):** The actual thing built from that blueprint (e.g., the physical house).
- **Example:**
  ```python
  class Example:
      pass

  a = Example()  # 'a' is an object made from the Example class
  ```

#### 2. Class Attributes vs. Instance Attributes
- **Class Attribute:** Defined directly inside the class. It is shared by **every single object** created from that class.
- **Instance Attribute:** Defined with `self.` inside methods. It is unique to **each specific object**.
- **Example:**
  ```python
  class Car:
      wheels = 4  # Class attribute: every car has 4 wheels

      def __init__(self, color):
          self.color = color  # Instance attribute: each car has its own color

  a1 = Car("red")
  a2 = Car("blue")
  print(a1.wheels)  # Output: 4
  print(a2.color)   # Output: blue
  ```

#### 3. The `__init__` Constructor and `self`
- **`__init__(self, ...)`:** The setup function that runs automatically the moment a new object is created.
- **`self`:** A required parameter that represents the specific object currently being created or used. It acts like a nametag telling Python whose data to access.

#### 4. Custom Methods
Methods are simply functions written inside a class that can read and modify the object's data. They always take `self` as their first parameter.
- **Example:**
  ```python
  def celebrate_birthday(self):
      self.age += 1
  ```

#### 5. Magic (Dunder) Methods
Python has special built-in methods with double underscores ("dunder") that define how objects behave with standard Python operations:
- **`__str__(self)`:** Controls what displays when you `print(object)`.
- **`__repr__(self)`:** Gives a detailed string meant for debugging.
- **`__len__(self)`:** Defines what `len(object)` returns.
- **`__eq__(self, other)`:** Defines what happens when using `==`.
- **Math operators:** `__add__` (`+`), `__sub__` (`-`), `__mul__` (`*`), `__truediv__` (`/`).

````python
# --- Demo: Classes, Attributes, and Methods ---

class Person:
    ID = 0  # Shared Class Attribute

    def __init__(self, name, age):
        # Instance attributes (unique to each person)
        self.name = name
        self.age = age
        Person.ID += 1
        self.ID = Person.ID
        self.friends = []

    def add_friend(self, friend_person):
        self.friends.append(friend_person)

    def __str__(self):
        return f"{self.name} (Age: {self.age}, ID: {self.ID})"

# Create two people
bob = Person("Bob Smith", 25)
alice = Person("Alice Parkson", 23)

bob.add_friend(alice)

print("Bob info:   ", bob)             # Uses __str__
print("Alice info: ", alice)           # Uses __str__
print("Bob's friend:", bob.friends[0].name)
````

## 8.2 Hidden Attributes and Methods (Encapsulation)

### Concept Summary
Encapsulation means protecting data inside an object so outside code does not accidentally overwrite or break it.

#### 1. Public Attributes (`self.name`)
- Regular variables with no underscores.
- Can be read and modified freely by any code anywhere.
- **Example:** `obj.name = "New Name"`

#### 2. Protected Attributes (`self._name`)
- Uses a **single underscore** prefix.
- It is a polite warning sign (🚧) to other programmers: *"This is meant for internal use or subclasses, please do not touch directly."*
- Python will still let you access it, but by convention you should not.
- **Example:** `self._branch = "North Branch"`

#### 3. Private Attributes (`self.__name`)
- Uses a **double underscore** prefix.
- Python actively renames the variable behind the scenes using **Name Mangling** to `_ClassName__attribute`.
- Trying to read `obj.__balance` from outside causes an `AttributeError`.
- **Example:** `self.__pin = 1234`

````python
# --- Demo: Public, Protected, and Private Attributes ---

class MyClass:
    def __init__(self):
        self.public_attr = "I'm public"
        self._protected_attr = "I'm protected"
        self.__private_attr = "I'm private"

obj = MyClass()

# 1. Public: Free to access
print("Public:   ", obj.public_attr)

# 2. Protected: Accessible, but convention says leave it alone
print("Protected:", obj._protected_attr)

# 3. Private: Trying direct access raises AttributeError
try:
    print(obj.__private_attr)
except AttributeError:
    print("Private:   Direct access blocked by Python!")

# 4. Name mangling (how Python secretly stores it)
print("Mangled:  ", obj._MyClass__private_attr)
````

### Exercise 2: Building a Secured Vault Account

**Requirements:**
1. Create a class named `VaultAccount`.
2. Inside `__init__(self, username, balance, pin)`:
   - Set `username` as public (`self.username`).
   - Set `balance` as protected with a single underscore (`self._balance`).
   - Set `pin` as private with a double underscore (`self.__pin`).
3. Add a method `verify_pin(self, entered_pin)` that returns `True` if `entered_pin == self.__pin`, otherwise `False`.
4. Create an account with `"Alice"`, balance `500`, and PIN `4321`.
5. Test accessing the private PIN inside a `try / except AttributeError` block to show direct access fails.
6. Test that `verify_pin(4321)` correctly returns `True`.

<details>
<summary>💡 Need a hint?</summary>

- Variable scopes: `self.username = username`, `self._balance = balance`, `self.__pin = pin`
- Verify method: `return entered_pin == self.__pin`
- Error test: `try: print(account.__pin) except AttributeError: print("Access blocked")`
</details>

````python
# --- Exercise 2: Your Code Here ---

class VaultAccount:
    def __init__(self, username, balance, pin):
        self.username = username
        self._balance = balance
        self.__pin = pin

    def verify_pin(self, entered_pin):
        return entered_pin == self.__pin

# Instantiate
account = VaultAccount("Alice", 500, 4321)

# Public access
print("Username:  ", account.username)

# Protected access (convention warns against direct use)
print("Balance:   ", account._balance)

# Private access check
try:
    print(account.__pin)
except AttributeError:
    print("Private PIN: Direct access blocked!")

# Safe check via method
print("PIN valid?: ", account.verify_pin(4321))

# Expected Output:
# Username:   Alice
# Balance:    500
# Private PIN: Direct access blocked!
# PIN valid?:  True
````

## 8.3 Inheritance and Polymorphism

### Concept Summary
Inheritance and Polymorphism allow classes to share logic and respond to the same action in their own specialized way.

#### 1. What is Inheritance?
- **Superclass (Parent / Base Class):** The general class containing common code.
- **Subclass (Child / Derived Class):** The specific class that inherits all code from the parent without rewriting it.
- **Syntax:** `class ChildClass(ParentClass):`

#### 2. The `super().__init__()` Function
- Instead of manually retyping attribute setup in the child, `super().__init__(...)` calls the parent's setup method directly.
- **Example:**
  ```python
  class Employee(Person):
      def __init__(self, name, age, position):
          super().__init__(name, age)  # Reuses parent Person setup
          self.position = position     # Adds employee-specific attribute
  ```

#### 3. What is Polymorphism?
- **Meaning:** "Many forms". It means different child classes can have methods with the exact same name, but each child performs that action in its own unique way.
- **Method Overriding:** When a child class defines a method that has the same name as the parent, the child's version replaces the parent's version.
- **Uniform Interface:** A single helper function can take any object and call that shared method without caring which exact class it came from.
- **Example:** Both a `Dog` and a `Cat` have a `.speak()` method, but one says `"Woof"` while the other says `"Meow"`.

````python
# --- Demo: Inheritance & Polymorphism ---

# Base Parent Class
class Animal:
    def speak(self):
        pass  # Placeholder

# Child Classes overriding speak()
class Dog(Animal):
    def speak(self):
        return "Woof Woof!"

class Cat(Animal):
    def speak(self):
        return "Meeow!"

class Duck(Animal):
    def speak(self):
        return "Quack!"

# Uniform function (Polymorphism in action)
def animal_sound(animal):
    return animal.speak()

# Testing with different animal objects
animals = [Dog(), Cat(), Duck()]
for a in animals:
    print(animal_sound(a))
````

### Exercise 3: Employee Hierarchy & Role Polymorphism

**Requirements:**
1. Create a parent class `Person`:
   - `__init__(self, name, age)` sets `self.name` and `self.age`.
   - Method `get_role(self)` returns `"General Person"`.
2. Create a child class `Employee` that inherits from `Person`:
   - `__init__(self, name, age, position)` uses `super().__init__(name, age)` to initialize the parent attributes, and sets `self.position`.
   - Override `get_role(self)` to return: `f"Employee working as {self.position}"`.
3. Add a helper method `last_name(self)` in `Employee` that returns only the last word of `self.name` (using `.split()[-1]`).
4. Create a `Person("Sam Green", 30)` and an `Employee("Jane Doe", 28, "Data Analyst")`.
5. Print `get_role()` for both objects and print Jane's extracted last name.

<details>
<summary>💡 Need a hint?</summary>

- Child class: `class Employee(Person):`
- Reuse parent setup: `super().__init__(name, age)`
- Overriding: Define `def get_role(self):` inside `Employee` with its own return string.
- Last name: `return self.name.split()[-1]`
</details>

````python
# --- Exercise 3: Your Code Here ---

# 1. Base Class
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def get_role(self):
        return "General Person"

# 2. Derived Subclass with super()
class Employee(Person):
    def __init__(self, name, age, position):
        super().__init__(name, age)
        self.position = position

    # 3. Method Overriding (Polymorphism)
    def get_role(self):
        return f"Employee working as {self.position}"

    def last_name(self):
        return self.name.split()[-1]

# Verification
person = Person("Sam Green", 30)
employee = Employee("Jane Doe", 28, "Data Analyst")

print("Person Role:   ", person.get_role())
print("Employee Role: ", employee.get_role())
print("Last Name:     ", employee.last_name())

# Expected Output:
# Person Role:    General Person
# Employee Role:  Employee working as Data Analyst
# Last Name:      Doe
````
