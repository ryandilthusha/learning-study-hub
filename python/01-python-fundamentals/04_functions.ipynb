## 4.1 Function Syntax, Return Values, and Flexible Arguments

### Concept Summary
A function is a named block of code that takes an input, does a job, and can give back a result. Turning messy or repeated code into clean functions is called **refactoring**.

#### 1. Defining and Returning (`def` and `return`)
- `def`: Tells Python you are starting a function.
- **Docstring (`'''...'''`)**: An optional text note right under the first line that describes what the function does.
- `return`: Sends the final calculated answer back to whoever called the function.
- **Example:**
  ```python
  def add_five(number):
      '''Adds 5 to any number'''
      return number + 5

  result = add_five(10)
  print(result)  # Output: 15
  ```

#### 2. Any Number of Inputs (`*args`)
If you do not know beforehand how many values someone will pass into your function, put `*args` as the parameter. Python packs all those extra values into a tuple.
- **Example:**
  ```python
  def add_all(*args):
      return sum(args)

  print(add_all(1, 2, 3, 4))  # Output: 10
  ```

#### 3. Any Number of Named Inputs (`**kwargs`)
If you want to pass named values of unspecified length, use `**kwargs`. Python packs them into a dictionary of key-value pairs.
- **Example:**
  ```python
  def show_profile(**kwargs):
      return kwargs

  print(show_profile(name="Astra", level=12))  # Output: {'name': 'Astra', 'level': 12}
  ```

````python
# --- Demo: Function Syntax & *args ---


# 1. Function with docstring and return
def adjusted_average(amount1, amount2):
    '''Calculates an adjusted average of two numbers'''
    average = (amount1 + amount2 + 1) / 2
    return average

ans = adjusted_average(2, 3)
print("Adjusted Average:", ans)                         # Output: 3.0
print("Docstring:", adjusted_average.__doc__)           # Output: Calculates an adjusted average of two numbers


# 2. Using *args to take any amount of numbers
def sum_all(*args):
    return sum(args)

total = sum_all(1, 2, 3, 4)
print("Sum of all numbers:", total)  # Output: 10
````

## 4.2 Recursion (A Function Calling Itself)

### Concept Summary
**Recursion** simply means a function that does a job by **calling itself** with a smaller input.

#### 1. Recursion vs. A Regular for Loop
Both do repeated work, but in different ways:

- **The `for` loop way (standard loop):**
  ```python
  for n in range(3, 0, -1):  # Counts down from 3 to 1
      print(n)
  # Outputs: 3, 2, 1
  ```

- **The recursion way (function calls itself):**
  ```python
  def count_down(n):
      if n == 0:
          return          # Stop when n reaches 0
      print(n)            # Print current number               # Outputs: 3 then 2 then 1
      count_down(n - 1)   # Call itself with a smaller number  

  count_down(3)           # Outputs: 3, 2, 1
  ```

#### 2. What Happens WITHOUT a Base Case? (The Crash)
If you do not tell the function when to stop, it calls itself forever until Python runs out of memory and crashes with an error (`RecursionError`).

```python
# ❌ DANGEROUS: No Base Case (Never stops!)
def bad_count(n):
    print(n)
    bad_count(n - 1)  # Keeps going: 3, 2, 1, 0, -1, -2, -3... CRASH!
```

#### 3. WITH a Base Case (The Stop Sign 🛑)
A **Base Case** is just an `if` check at the start of your function that says: *"If we reach this number, stop and do not call yourself again."*

```python
# ✅ SAFE: Has a Base Case
def good_count(n):
    # 1. BASE CASE (The Stop Sign):
    if n == 0:
        print("Done!")
        return            # Stops the function right here

    # 2. RECURSIVE STEP (One step smaller):
    print(n)
    good_count(n - 1)     # Calls itself with n - 1
```

````python
# --- Demo: Factorial using Recursion ---

def factorial(x):
    # 1. BASE CASE: Stop when x is 1
    if x == 1:
        return 1
    
    # 2. RECURSIVE CASE: Call itself with a smaller number (x - 1)
    else:
        return x * factorial(x - 1)

# --- How factorial(3) breaks down across calls ---
# Expanding down (call stack):
#   factorial(3) = 3 * factorial(2)
#                = 3 * (2 * factorial(1))
#                = 3 * (2 * 1)            <- Base case reached!

# Collapsing back up (returns):
#                = 3 * 2
#                = 6                      3 * 2 * 1                                                                                                 


ans = factorial(3)  # 3 * 2 * 1
print("Factorial of 3 is:", ans)  # Output: 6
````

## 4.3 Common Function Issues (Scope & Default Arguments)

### Concept Summary

#### 1. Global vs. Local Variables
- **Global Variable:** Created outside of any function. Anyone can read it.
- **Local Variable:** Created inside a function. It lives only while the function is running.
- Changing a variable inside a function does **not** change the global variable outside.
  - **Example:**
    ```python
    x = 5
    def test():
        x = 99  # this is a brand new local x
    test()
    print(x)    # Output: 5 (the global x is untouched)
    ```

#### 2. Modifying Lists Inside Functions
Even though you cannot reassign numbers easily, modifying a list directly (like `.append()`) **will** change the original list outside!
- **Example:**
  ```python
  def add_item(my_list):
      my_list.append(4)

  nums = [1, 2, 3]
  add_item(nums)
  print(nums)  # Output: [1, 2, 3, 4]
  ```

#### 3. The `global` Keyword vs. Best Practice
- The `global` keyword forces a function to overwrite an outside variable directly.
- **Best Practice:** Do **not** use `global`. Instead, pass the value into the function, modify it locally, and `return` the new value.

#### 4. The Mutable Default Argument Trap
Default values like `number_list=[]` are created **only once** when Python first reads the function definition, not each time the function is called.
- If you add items to a default list, it keeps growing across future function calls!
- **Example:**
  ```python
  def add_number(num, num_list=[]):
      num_list.append(num)
      return num_list

  print(add_number(5))  # Output: [5]
  print(add_number(7))  # Output: [5, 7]  (reused the same list!)
  ```

````python
# --- Demo: Scope & The Default List Trap ---

# 1. Local vs Global
x = 5
def bad_function(value):
    x = value
    return x

print("Function returns:", bad_function(10))  # Output: 10
print("Global x stays:  ", x)                 # Output: 5

# 2. Modifying a list in-place
def modify_list(L):
    L.append(4)

original_list = [1, 2, 3]
modify_list(original_list)
print("Modified List:   ", original_list)     # Output: [1, 2, 3, 4]

# 3. The Mutable Default Argument Trap
def append_item(number, number_list=[]):
    number_list.append(number)
    return number_list

print("Call 1:", append_item(5))  # Output: [5]
print("Call 2:", append_item(7))  # Output: [5, 7]
print("Call 3:", append_item(2))  # Output: [5, 7, 2]
````
