## 2.1 Strings

### Concept Summary
Strings are ordered, immutable sequences of text characters.

#### 1. Concatenation (`+`)
Combines two strings into one.
- **Example:** `"Data" + "Science"` $\rightarrow$ `'DataScience'`

#### 2. Indexing (`string[index]`)
Grabs a single character by its 0-based position (negative numbers count from the end).
- **Example:** `"apple"[1]` $\rightarrow$ `'p'`
- **Example:** `"apple"[-1]` $\rightarrow$ `'e'`

#### 3. Slicing (`string[start:stop]`)
Extracts a range of characters from `start` up to, but not including, `stop`.
- **Example:** `"apple"[1:4]` $\rightarrow$ `'ppl'`
- **Example:** `"banana"[:3]` $\rightarrow$ `'ban'`

#### 4. Step Slicing (`string[start:stop:step]`)
Skips characters by a specified step size.
- **Example:** `"apple"[::2]` $\rightarrow$ `'ape'` (every 2nd character)

#### 5. Reversing a String (`string[::-1]`)
Uses a negative step size of `-1` to reverse the sequence.
- **Example:** `"apple"[::-1]` $\rightarrow$ `'elppa'`

#### 6. Splitting (`str.split(sep)`)
Splits a string by a delimiter and returns a list of pieces (defaults to whitespace).
- **Example:** `"Hello World".split()` $\rightarrow$ `['Hello', 'World']`
- **Example:** `"A,B,C".split(",")` $\rightarrow$ `['A', 'B', 'C']`

#### 7. Joining (`sep.join(list_of_strings)`)
Joins a list of strings into one string with a custom glue in between.
- **Example:** `" ".join(['Hello', 'World'])` $\rightarrow$ `'Hello World'`
- **Example:** `"-".join(['2026', '10', '23'])` $\rightarrow$ `'2026-10-23'`

#### 8. Stripping (`str.strip(chars)`)
Removes leading and trailing characters (defaults to whitespace).
- **Example:** `"  clean me  ".strip()` $\rightarrow$ `'clean me'`
- **Example:** `"xxPythonxx".strip("x")` $\rightarrow$ `'Python'`

```python
# --- Demo: String Operations ---
sample = "  apple,banana,cherry  "

# 1. Strip whitespace
cleaned = sample.strip()
print("Cleaned:", cleaned)      #Output: Cleaned: apple,banana,cherry

# 2. Split into list
fruits = cleaned.split(",")
print("Fruits List:", fruits)   #Output: Fruits List: ['apple', 'banana', 'cherry']

# 3. Rejoin with arrows
joined_text = " -> ".join(fruits)
print("Joined:", joined_text)   #Output: Joined: apple -> banana -> cherry

# 4. Slice & Reverse
first_fruit = fruits[0]
print("First Fruit:", first_fruit)      #Output: First Fruit: apple
print("Reversed:", first_fruit[::-1])   #Output: Reversed: elppa
```

## 2.2 Lists

### Concept Summary
Lists are mutable ordered sequences that can hold mixed data types.

#### 1. Adding Elements
- `list.append(item)`: Adds `item` to the very end.
  - **Example:** `nums = [1, 2]; nums.append(3)` $\rightarrow$ `nums` is now `[1, 2, 3]`
- `list.insert(index, item)`: Inserts `item` at a specific position.
  - **Example:** `nums = [1, 3]; nums.insert(1, 2)` $\rightarrow$ `nums` is now `[1, 2, 3]`

#### 2. Removing Elements
- `list.remove(val)`: Deletes the **first occurrence** of `val`.
  - **Example:** `letters = ['a', 'b', 'a']; letters.remove('a')` $\rightarrow$ `letters` is now `['b', 'a']`
- `list.pop(index)`: Deletes and returns the element at `index` (defaults to the last element).
  - **Example:** `items = [10, 20, 30]; val = items.pop(1)` $\rightarrow$ `val` is `20`, `items` is now `[10, 30]`

#### 3. Overwriting Values (`list[index] = new_value`)
Directly replaces an element in-place.
- **Example:** `vals = [10, 99]; vals[1] = 20` $\rightarrow$ `vals` is now `[10, 20]`

#### 4. Sorting Lists: `sorted()` vs `.sort()`
- `sorted(list)`: Returns a **new** sorted list. The original list is **not modified**.
  - **Example:** `a = [3, 1]; b = sorted(a)` $\rightarrow$ `b` is `[1, 3]`, `a` remains `[3, 1]`
- `list.sort()`: Modifies the original list **in-place** and returns `None`.
  - **Example:** `a = [3, 1]; a.sort()` $\rightarrow$ `a` is modified to `[1, 3]`
- Use `reverse=True` inside either function to sort in descending order:
  - **Example:** `sorted([1, 4, 2], reverse=True)` $\rightarrow$ `[4, 2, 1]`

#### 5. List Comprehension
A compact syntax for creating a new list by transforming or filtering elements.
- **Format:** `[expression for item in iterable if condition]`
- **Example (Transform):** `[x * 2 for x in [1, 2, 3]]` $\rightarrow$ `[2, 4, 6]`
- **Example (Filter):** `[x for x in range(6) if x % 2 == 0]` $\rightarrow$ `[0, 2, 4]`

```python
# --- Demo: List Operations & Comprehensions ---
nums = [3, 1, 4, 1, 5]

# In-place addition & removal
nums.append(9)
nums.insert(0, 0)
nums.remove(1)      # removes first occurrence of 1
popped = nums.pop() # removes 9 from the end
print("Modified nums:", nums)
print("Popped item:  ", popped)

# Non-mutating sort
print("Sorted copy:  ", sorted(nums, reverse=True))
print("Original:     ", nums)

# List Comprehension
doubled_evens = [x * 2 for x in nums if x % 2 == 0]
print("Doubled evens:", doubled_evens)
```

## 2.3 Sets

### Concept Summary with Examples
Sets are unordered collections of unique, hashable elements based on hash tables.

#### 1. Initialization & Deduplication
- Empty set: `my_set = set()` (Note: `{}` creates an empty dictionary!)
- Deduplication from list: `set(iterable)` strips duplicate items automatically.
  - **Example:** `set([1, 1, 2, 3, 3])` $\rightarrow$ `{1, 2, 3}`

#### 2. Adding Elements (`set.add(item)`)
Adds an item to the set if it is not already present.
- **Example:** `s = {1, 2}; s.add(3)` $\rightarrow$ `s` is `{1, 2, 3}`
- **Example:** `s.add(2)` $\rightarrow$ does nothing because `2` is already in `s`.

#### 3. Removing Elements (`set.remove(item)`)
Deletes an item from the set. Raises a `KeyError` if the item is not found.
- **Example:** `s = {1, 2, 3}; s.remove(2)` $\rightarrow$ `s` is `{1, 3}`

#### 4. Fast Membership Testing (`item in set`)
Because sets use hash tables, checking membership is instantaneous ($O(1)$ time complexity), far faster than searching a list.
- **Example:** `"apple" in {"apple", "banana"}` $\rightarrow$ `True`
- **Example:** `"grape" in {"apple", "banana"}` $\rightarrow$ `False`

#### 5. Sets Are Unordered & Non-Indexable
Sets do not maintain positions or sequential indices.
- **Example:** `{1, 2, 3}[0]` $\rightarrow$ raises `TypeError: 'set' object is not subscriptable`.

#### 6. Only Immutable (Hashable) Items Allowed
You can store numbers, strings, and tuples in a set. Adding mutable items like lists or dicts fails.
- **Example:** `{ [1, 2] }` $\rightarrow$ raises `TypeError: unhashable type: 'list'`.

```python
# --- Demo: Set Operations ---
numbers = [1, 2, 2, 3, 4, 4, 4]

# 1. Deduplicate
unique = set(numbers)
print("Unique Set:", unique)

# 2. Add and Remove
unique.add(10)
unique.remove(2)
print("After Add/Remove:", unique)

# 3. Membership test
print("Is 10 present?:", 10 in unique)
print("Is 99 present?:", 99 in unique)
```

## 2.4 Dictionaries

### Concept Summary with Examples
Dictionaries store data as `key: value` pairs. Keys must be unique and immutable (hashable), while values can be any Python data type.

#### 1. Initializing a Dictionary
- Literal syntax: `d = {'one': 1, 'two': 2}`
- Constructor syntax: `d = dict(one=1, two=2)`

#### 2. Accessing & Adding/Updating Key-Value Pairs
- Read: `my_dict[key]`
  - **Example:** `user = {'name': 'Sam'}; print(user['name'])` $\rightarrow$ `'Sam'`
- Add or Update: `my_dict[key] = new_value`
  - **Example:** `user['age'] = 25` $\rightarrow$ adds `'age': 25`
  - **Example:** `user['name'] = 'Alex'` $\rightarrow$ updates `'name'` to `'Alex'`

#### 3. Dictionary Views (`.keys()`, `.values()`, `.items()`)
- `dict.keys()`: An iterable view of all keys.
  - **Example:** `list({'a': 1, 'b': 2}.keys())` $\rightarrow$ `['a', 'b']`
- `dict.values()`: An iterable view of all values.
  - **Example:** `list({'a': 1, 'b': 2}.values())` $\rightarrow$ `[1, 2]`
- `dict.items()`: An iterable view of `(key, value)` tuples.
  - **Example:** `list({'a': 1, 'b': 2}.items())` $\rightarrow$ `[('a', 1), ('b', 2)]`

#### 4. Removing Keys (`dict.pop(key)`)
Deletes `key` from the dictionary and returns its `value`.
- **Example:** `scores = {'math': 90, 'chem': 80}; removed = scores.pop('chem')` $\rightarrow$ `removed` is `80`, `scores` is now `{'math': 90}`

```python
# --- Demo: Dictionary Operations ---
student = {"name": "Jordan", "track": "Data Analytics", "score": 85}

# Add and Update
student["status"] = "Active"
student["score"] += 5

# Pop a key
removed_track = student.pop("track")
print("Popped Track:", removed_track)

# Iterate through key-value pairs
print("\nRemaining Student Details:")
for key, val in student.items():
    print(f"  - {key}: {val}")
```
