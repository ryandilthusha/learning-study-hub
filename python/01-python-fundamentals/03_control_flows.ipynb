## 3.1 Conditional Statements

### Concept Summary
Conditional statements run specific blocks of code based on whether a condition evaluates to `True`.

#### 1. The `if` Statement
Runs the indented code block only if its condition evaluates to `True`. It can stand completely on its own.
- **Example:**
  ```python
  x = 10
  if x > 5:
      print("Greater")  # Output: Greater
  ```

#### 2. The `elif` Statement
Short for "else if". Checks another condition only if the previous `if` or `elif` condition was `False`. You can have one or many `elif` blocks, but they must come after an `if` and before `else`.
- **Example:**
  ```python
  score = 75
  if score >= 90:
      print("A")
  elif score >= 70:
      print("B")  # Output: B
  ```

#### 3. The `else` Statement
Executes only if **all** preceding `if` and `elif` conditions were `False`.
- **Example:**
  ```python
  val = 3
  if val > 5:
      print("High")
  else:
      print("Low")  # Output: Low
  ```

#### 4. Boolean Conditions (`True`, `False`, `None`)
An `if` statement only checks if something is **True (Yes)** or **False (No)**:

- **Comparing numbers:**
  - `10 > 5` $\rightarrow$ `True` (Yes, 10 is bigger than 5)
  - `4 == 9` $\rightarrow$ `False` (No, 4 does not equal 9)

- **Checking inside a list (`in`):**
  - `'apple' in ['apple', 'banana']` $\rightarrow$ `True` (Yes, apple is there)
  - `'milk' in ['apple', 'banana']` $\rightarrow$ `False` (No, milk is not there)

- **What about `None`?**
  - `None` means empty or nothing. Python treats `None` as `False`.

#### 5. Order & Nesting (The Rules of if, elif, else)

**Rule 1: `if` can stand completely alone**
You do not need to write `elif` or `else`. A single `if` works fine by itself:
```python
age = 20
if age >= 18:
    print("Welcome!")  # Runs if age is 18 or older
```

**Rule 2: `elif` and `else` always need an `if` first**
You cannot start with `elif` or `else`. They must follow an `if`:
- **`if`** goes first.
- **`elif`** goes in the middle.
- **`else`** goes at the end.

**Rule 3: Nesting (`if` inside another `if`)**
You can put an `if` inside another `if`. The inner test only runs if the first test is `True`:
```python
has_ticket = True
age = 25

if has_ticket:
    if age >= 18:
        print("Allowed to enter!")  # Only runs if BOTH tests pass
```


#### 6. The `pass` Statement (Do Nothing / Skip for Now)
In Python, you cannot leave an `if` block empty. If you do, Python throws an error and crashes.

If you want to write an `if` statement but are not ready to add code inside it yet, write **`pass`**.

- **What it means:** "Do nothing for now, just keep going."
- **Why use it:** It keeps Python from crashing while you build your code.

```python
score = 90

if score > 50:
    pass  # Will write code here later, do nothing right now
```

````python
# --- Demo: Conditional Statements ---
num = 15

# if - elif - else chain
if num > 20:
    print("Number is large")
elif num > 10:
    print("Number is medium")   # Output: Number is medium
else:
    print("Number is small")

# Nested condition with membership check and pass
fruits = ["apple", "banana"]
selected = "apple"

if selected in fruits:
    if selected == "banana":
        pass  # placeholder
    else:
        print("Selected apple") # Output: Selected apple
````

## 3.2 For Loops

### Concept Summary
A `for` loop iterates over a sequence (such as a list, string, tuple, or range) and runs the code block once for each item.

#### 1. Basic Sequence Iteration
Stores the current element into a loop variable during each cycle.
- **Example:**
  ```python
  for x in ["a", 2, "school"]:
      print(x)
  # Outputs: "a", 2, "school"
  ```

#### 2. The `continue` Statement
Skips the rest of the current iteration and jumps directly to the next item in the sequence.
- **Example:**
  ```python
  for num in [1, 2, 3]:
      if num == 2:
          continue
      print(num)  # Outputs: 1, 3
  ```

#### 3. The `break` Statement
Stops the loop completely and exits immediately, skipping all future iterations.
- **Example:**
  ```python
  for num in [1, 2, 3]:
      if num == 2:
          break
      print(num)  # Outputs: 1
  ```

````python
# --- Demo: For Loops with continue and break ---
sequence = ["a", 2, "school"]

# Loop with continue (skips 2)
print("Using continue:")
for data in sequence:
    if data == 2:
        continue
    print(data)         # Output: a, school

print("")

# Loop with break (stops at 2)
print("Using break:")
for data in sequence:
    if data == 2:
        break
    print(data)         # Output: a
````

## 3.3 While Loops

### Concept Summary
A `while` loop repeatedly executes its code block as long as its condition remains `True`.

#### 1. Basic Structure
Runs repeatedly until the test condition evaluates to `False`.
- **Example:**
  ```python
  while condition:
      pass
  ```

#### 2. Loop Termination & Counter Updates
To prevent infinite loops, the program must have a way to exit:
- By updating a variable so the condition becomes `False` (e.g., `count += 1`).
- Or by triggering a `break` statement inside the loop.
- `pass`, `continue`, and `break` function the same way as they do in `for` loops.
- **Example:**
  ```python
  count = 1
  while count <= 3:
      print(count)
      count += 1
  # Outputs: 1, 2, 3
  ```

````python
# --- Demo: While Loop Counter ---
count = 1

while count <= 5:
    print("Count is:", count)
    count += 1

print("Loop ended!")

# Output:
# Count is: 1
# Count is: 2
# Count is: 3
# Count is: 4
# Count is: 5
# Loop ended!
````
