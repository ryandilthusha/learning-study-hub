## 5.1 Inputs in Python

### Concept Summary
The `input()` function lets a user type information into your program from the keyboard.

#### 1. How `input()` Works
You place a prompt message inside the parentheses to tell the user what to type.
- **Example:**
  ```python
  name = input("Enter your name: ")
  print("Hello " + name)

  # Output:
  # Enter your name: Alice
  ```

#### 2. `input()` Always Returns Text (String)
No matter what the user types, Python always saves it as a string (`str`). Even if they type the number `3`, Python sees `"3"`, not the number `3`.
- **Example:**
  ```python
  val = input("Give me a number: ")  # User types 3
  print(type(val))                   # Output: <class 'str'>
  ```

#### 3. Converting Inputs for Math & Handling Edge Cases
If you need to do math with an input, you must convert it using `int()` or `float()`. Since users might type unexpected values, use `if` / `else` checks or test functions to prevent your code from crashing.
- **Example:**
  ```python
  age_text = input("Enter age: ")
  age = int(age_text)  # Converts "20" -> 20
  print(age + 1)
  ```

````python
# --- Demo: Working with input() ---

# Simulating user input:
data = "3"
print("Raw input value:      ", data)                 # Output: 3
print("Data type of input:   ", type(data))           # Output: <class 'str'>

# Converting to integer for math:
number = int(data)
print("Math result (3 + 5):  ", number + 5)           # Output: 8
````

## 5.2 Type Conversion & Number Rounding

### Concept Summary
Converting types means changing a value from one data type to another. Trying to convert incompatible types causes a `TypeError` or `ValueError`.

#### 1. Converting Numbers and Text
- **`int(x)`**: Converts to a whole number. When converting floats, it cuts off the decimal part (floor).
  - `int(3.9)` $\rightarrow$ `3`
  - `int("15")` $\rightarrow$ `15`
- **`float(x)`**: Converts to a decimal number.
  - `float(4)` $\rightarrow$ `4.0`
  - `float("3.14")` $\rightarrow$ `3.14`
- **`str(x)`**: Turns any value into readable text.
  - `str(10)` $\rightarrow$ `'10'`
  - `str([1, 2])` $\rightarrow$ `'[1, 2]'`

#### 2. Converting Collections
- **`list(str)`**: Splits a word into a list of characters: `list("cat")` $\rightarrow$ `['c', 'a', 't']`
- **`set(list)`**: Removes duplicates: `set([1, 2, 2])` $\rightarrow$ `{1, 2}`
- **`dict.fromkeys(keys)`**: Builds a dictionary from a list or set of keys: `dict.fromkeys(["a", "b"])` $\rightarrow$ `{'a': None, 'b': None}`

#### 3. Decimal Rounding with `:.2f`
The `f` specifier sets the exact number of decimal places to show.
- `f"{3.14159:.2f}"` $\rightarrow$ `'3.14'` (2 decimal places)
- `f"{3.14159:.4f}"` $\rightarrow$ `'3.1416'` (4 decimal places)

#### 4. Significant Figures with `:.3g`
The `g` specifier (general format) keeps a fixed number of important digits (significant figures). For very large or very tiny numbers, it automatically switches to scientific notation (`e`).
- `f"{123456789:.3g}"` $\rightarrow$ `'1.23e+08'` (keeps 3 main digits)
- `f"{0.0000123456:.3g}"` $\rightarrow$ `'1.23e-05'` (keeps 3 main digits)

````python
# --- Demo: Type Conversion & Rounding ---

# Basic conversions
print("Float to Int (drops decimal):", int(8.95))             # Output: 8
print("Int to Float:                 ", float(10))              # Output: 10.0
print("String to List:               ", list("code"))           # Output: ['c', 'o', 'd', 'e']
print("List to Dict keys:            ", dict.fromkeys(["x", "y"])) # Output: {'x': None, 'y': None}

# Formatting numbers
pi = 3.14159265
big_num = 123456789
small_num = 0.0000123456

print(f"\n2 Decimal Places: {pi:.2f}")                         # Output: 3.14
print(f"Big number (3 sig figs):   {big_num:.3g}")           # Output: 1.23e+08
print(f"Small number (3 sig figs): {small_num:.3g}")         # Output: 1.23e-05
````

## 5.3 Print Statement Formatting

### Concept Summary
Formatting controls how text looks on the screen using f-strings, alignment spaces, and line breaks.

#### 1. Using f-Strings
Put an `f` before your quotes and place variables inside curly brackets `{}`.
- **Example:**
  ```python
  score = 95
  print(f"Your score is {score}")  # Output: Your score is 95
  ```

#### 2. Spacing and Alignment (`<` and `>`)
You can pad text with spaces to make clean tables:
- **`{text:<10}`**: Left-aligns text and pads spaces to the right (total width of 10).
- **`{text:>10}`**: Right-aligns text and pads spaces to the left (total width of 10).
- **Example:** `f"{item:<10} {price:>6}"`
- **Output:**
  ```
  Apple      1.50
  Banana     0.75
  ```


#### 3. Tabs (`\t`) and New Lines (`\n`)
- **`\t`**: Inserts a tab space (equal to 4 spaces).
- **`\n`**: Moves text to the next line.
- **Example:** `print("First Line\n\tIndented Second Line")`
- **Output:**
  ```
  First Line
      Indented Second Line
  ```

#### 4. The `end` Parameter in `print()`
By default, `print()` ends with a new line (`\n`). You can change this using `end="..."` so the next print stays on the same line.
- **Example:**
  ```python
  print("Hello", end=" ")
  print("World")  # Output: Hello World (on one single line)
  ```

````python
# --- Demo: Spacing, Alignment & print() options ---


# 1. Padding with < and >
print(f"{'Product':<12} | {'Cost':>8}")
print("-" * 23)
print(f"{'Keyboard':<12} | {'$45.00':>8}")
print(f"{'Mouse':<12} | {'$12.50':>8}")


# 2. Tabs and Newlines
print("\nItem List:\n\t* Pen\n\t* Notebook")


# 3. Changing print end
print("\nCounting: ", end="")
for n in [1, 2, 3]:
    print(n, end=" -> ")
print("Finish!")
````
