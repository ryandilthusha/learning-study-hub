## Part 1: Augmented Assignment Operators

### The Concept
In programming, you often want to take a variable, modify its current value, and store the result back into the same variable.

Instead of writing the variable name twice:
```python
a = a + 1
```

Python provides an **augmented assignment shorthand**:
```python
a += 1
```

### Common Operators
| Shorthand | Equivalent Longhand | Operation |
| :--- | :--- | :--- |
| `a += b` | `a = a + b` | Addition |
| `a -= b` | `a = a - b` | Subtraction |
| `a *= b` | `a = a * b` | Multiplication |
| `a /= b` | `a = a / b` | Float Division |
| `a //= b` | `a = a // b` | Integer (Floor) Division |
| `a %= b` | `a = a % b` | Modulo (Remainder) |
| `a **= b` | `a = a ** b` | Exponentiation (Power) |

> **Note:** Shorthand assignment works for arithmetic operations, but not for boolean comparisons like `==` or `!=`.

````python
# --- Concept Demo: Augmented Assignment ---

counter = 10
counter += 5   # Same as: counter = counter + 5 -> 15
counter -= 3   # Same as: counter = counter - 3 -> 12
counter *= 2   # Same as: counter = counter * 2 -> 24
counter //= 4  # Same as: counter = counter // 4 -> 6

print("Final counter value:", counter)
````

## Part 2: Multiple Assignment & Sequence Unpacking

### The Concept
Python lets you assign multiple variables simultaneously in a single line. This keeps your code concise and readable.



#### 1. Assigning Values Directly in One Line
```python
a, b = 1, 2
# a is 1, b is 2
```



#### 2. Unpacking a List or Tuple
If you have a collection (list or tuple), you can unpack its items directly into individual variables:
```python
coordinates = [10, 20]
x, y = coordinates
```



#### 3. Unpacking Function Returns
Functions in Python can return multiple values separated by commas. You can unpack them cleanly in a single line:
```python
def get_user():
    return "Alice", 25

name, age = get_user()
```

````python
# --- Concept Demo: Multiple Assignment & Unpacking ---

# Direct multi-assignment
first, second, third = 10, 20, 30
print("Direct:", first, second, third)



# Unpacking a list
rgb_values = [255, 128, 0]
red, green, blue = rgb_values
print(f"RGB: R={red}, G={green}, B={blue}")



# Unpacking function return values
def get_circle_info():
    radius = 5
    area = 78.54
    return radius, area

r, a = get_circle_info()
print(f"Circle: radius={r}, area={a}")
````
