## Quick Guide: Which Mean Should You Use?

In statistics, "average" does not always mean adding numbers and dividing by the count. Picking the wrong type of mean can give a false answer:

| Type of Mean | How It Works | Best For | Example |
| :--- | :--- | :--- | :--- |
| **Arithmetic Mean** | Add all numbers, divide by how many there are | Normal measurements where every item matters equally (heights, test scores) | 5 students scored 80, 90, 70, 85, and 95 on a test. The arithmetic mean is (80 + 90 + 70 + 85 + 95) / 5 = 84. |
| **Geometric Mean** | Multiply all numbers together, then take the $n$-th root | Growth rates and percentages (investment returns, population growth) | Ex: 3 years of investment returns: 10%, 20%, and 30%. The geometric mean is $\sqrt[3]{(1.10)(1.20)(1.30)} - 1 \approx 19.7\%$. |
| **Harmonic Mean** | Flip numbers into fractions (`1 / x`), average them, and flip back | Rates and speeds over a fixed distance (car speeds in `km/h`, price ratios) | Ex: A car travels 60 km at 60 km/h and returns 60 km at 30 km/h. The harmonic mean is $2 / ((1/60) + (1/30)) = 40$ km/h. |
| **Weighted Mean** | Multiply each number by its importance (weight), then divide by total weight | Data where some items matter more than others (exam weights, big vs. small surveys) | Ex: A student has 2 exams, one worth 40% and the other worth 60%. They scored 80 on the first and 90 on the second. The weighted mean is (0.4 * 80 + 0.6 * 90) / (0.4 + 0.6) = 86. |

## 3.1 Arithmetic Mean

### Concept Summary
The **Arithmetic Mean** is the everyday "average" most people know. It works well when every data point is equally important.

#### Formula
$$A = \frac{1}{n} \sum_{i=1}^{n} a_i = \frac{a_1 + a_2 + \dots + a_n}{n}$$
- $n$ = how many numbers you have
- $a_i$ = each individual number
- $\sum$ = "sum" (add everything up)

**Step-by-Step Example:**
Suppose 5 friends have heights of `186 cm`, `179 cm`, `182 cm`, `165 cm`, and `173 cm`:
1. Add them up: $186 + 179 + 182 + 165 + 173 = 885$
2. Divide by 5 friends: $\frac{885}{5} = \mathbf{177\text{ cm}}$

#### The Outlier Problem
The arithmetic mean is easily fooled by **outliers** (a few numbers that are way bigger or smaller than the rest):
- In 2022, the arithmetic mean US household income was **105,555 dollars**, while the median (exact middle household) was **74,580 dollars**.
- Why the huge gap? A small number of billionaires pulled the arithmetic mean way higher than what a normal family actually earns.

### How to Code Arithmetic Mean in Python

#### 1. Normal Python (Lists)
- **`sum(data) / len(data)`:** Simple, clean, and fast for normal Python lists.
- **`statistics.mean(data)`:** Built-in module (`import statistics`). A bit slower, but handles special number types like fractions accurately.
- **`statistics.fmean(data)`:** Converts numbers to decimals (`float`) first, making it faster than `statistics.mean()`.

#### 2. NumPy (`np.mean`)
For large arrays or 2D number grids, NumPy is much faster than normal Python and lets you pick columns or rows using `axis`:
```python
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

np.mean(matrix)          # Average of all 6 numbers     -> 35.0
np.mean(matrix, axis=0)  # Down columns (Column means)  -> [25. 35. 45.]
np.mean(matrix, axis=1)  # Across rows (Row means)      -> [20. 50.]
```

#### 3. Pandas DataFrames (`df.mean()`)
In Pandas, `.mean(axis=0, skipna=True)` automatically ignores `NaN` (missing cells) so empty spots do not ruin your count $n$:
```python
df = pd.DataFrame({'a': [1, 2], 'b': [5, 3]}, index=['tiger', 'zebra'])

# Output DataFrame:
#        a  b
# tiger  1  5
# zebra  2  3

df.mean(axis=0)  # Column means -> a: 1.5, b: 4.0
df.mean(axis=1)  # Row means    -> tiger: 3.0, zebra: 2.5
```

````python
# --- Demo: Arithmetic Mean & Speed Check ---
import statistics
import time
import numpy as np
import pandas as pd


# 1. Speed test on 1,000,000 numbers
print("----- Demo Question 1 -----")
np.random.seed(42)
arr_data = np.random.uniform(0, 100, 1_000_000)     # 1 million random numbers
list_data = list(arr_data)

start = time.time()
py_mean = sum(list_data) / len(list_data)
py_time = time.time() - start

start = time.time()
stat_mean = statistics.fmean(list_data)
stat_time = time.time() - start

start = time.time()
np_mean = np.mean(arr_data)
np_time = time.time() - start

print(f"Python sum()/len():   {py_mean:.4f} | Time: {py_time:.5f}s")
print(f"statistics.fmean():   {stat_mean:.4f} | Time: {stat_time:.5f}s")
print(f"NumPy np.mean():      {np_mean:.4f} | Time: {np_time:.5f}s")



# 2. Pandas DataFrame row & column means (skips NaN automatically)
print("\n----- Demo Question 2 -----")
df = pd.DataFrame({
    'Math':    [80, 90, np.nan],
    'Physics': [70, 85, 95]
}, index=['Alice', 'Bob', 'Charlie'])

print("=== DataFrame ===")
print(df)

print("\n=== Column Means ===")
print(df.mean(axis=0))

print("\n=== Row Means ===")
print(df.mean(axis=1))
````

## 3.2 Geometric Mean

### Concept Summary
When data changes by **multiplying** step-by-step—like investment returns, interest rates, or growing populations—the normal arithmetic mean gives the wrong answer. We use the **Geometric Mean** instead.

#### Step-by-Step Example: Why Arithmetic Mean Fails on Money
Suppose you start with **1,000 dollars** in stocks. Over 5 years, your yearly returns are: `+10%`, `-12%`, `+90%`, `-30%`, and `+25%`.

**1. What REALLY happens to your 1,000 dollars?**
First, convert each percentage into a **growth multiplier** (start at `1.00`, add gains, subtract losses):
- `+10%` gain $\rightarrow$ `1 + 0.10 = 1.10`
- `-12%` loss $\rightarrow$ `1 - 0.12 = 0.88`
- `+90%` gain $\rightarrow$ `1 + 0.90 = 1.90`
- `-30%` loss $\rightarrow$ `1 - 0.30 = 0.70`
- `+25%` gain $\rightarrow$ `1 + 0.25 = 1.25`

Now multiply your 1,000 dollars year by year (**compounding** = applying each new percentage change on top of the previous year's new total):
- Year 1: $1,000 \times 1.10 = 1,100$
- Year 2: $1,100 \times 0.88 = 968$
- Year 3: $968 \times 1.90 = 1,839.20$
- Year 4: $1,839.20 \times 0.70 = 1,287.44$
- Year 5: $1,287.44 \times 1.25 = \mathbf{1,609.30}$ (about **1,609 dollars**)

**2. Why the Normal Arithmetic Mean is WRONG:**
- If you average the 5 percentages the normal way:
  $$\frac{10\% + (-12\%) + 90\% + (-30\%) + 25\%}{5} = \frac{83\%}{5} = \mathbf{16.6\%}$$
- If your 1,000 dollars grew by `+16.6%` (`1.166`) every year for 5 years:
  $$1,000 \times 1.166 \times 1.166 \times 1.166 \times 1.166 \times 1.166 = 1,000 \times (1.166)^5 = \mathbf{2,155}$$
- That predicts **2,155 dollars**, which is way off from the real **1,609 dollars**!

**3. How the Geometric Mean gives the RIGHT Answer:**
Instead of adding, we **multiply** all 5 multipliers together, then take the **5th root** (raising to the power `1/5`):
1. Multiply all 5 multipliers: $1.10 \times 0.88 \times 1.90 \times 0.70 \times 1.25 = 1.6093$
2. Take the 5th root: $(1.6093)^{1/5} = \mathbf{1.0998}$ `This is Geometric Mean  (the average growth multiplier per year)`
3. Check the math for 5 years: $1,000 \times (1.0998)^5 = \mathbf{1,609.30}$ — it lands right on the real final amount!

Subtract `(1.0998 - 1)  = 0.0998`, which means your money grew by an average of **9.98% per year**.

#### Formula
$$G = \left( \prod_{i=1}^{n} x_i \right)^{\frac{1}{n}} = \sqrt[n]{x_1 \cdot x_2 \cdots x_n}$$
- $\prod$ = "product" (multiply all numbers together)
- Taking the $n$-th root in Python is written as `**(1 / n)`

#### Limitations
- **No negative numbers:** Taking roots of negative numbers gives imaginary/complex numbers (`NaN`). Always turn percentage losses like `-12%` into positive multipliers (`0.88`) first!
- **No zeros:** If any value is `0`, multiplying everything together becomes `0`.

---

### How to Code Geometric Mean in NumPy
Use `np.prod()` to multiply all items in an array together, then raise to the power `1 / n`:
```python
import numpy as np

def geometric_mean(data, axis=None):
    arr = np.array(data)

    if arr.ndim > 1 and axis is not None:     # If this is 2D AND have mention an axis
        n = arr.shape[axis]
        return np.prod(arr, axis=axis) ** (1 / n)
    else:                                     # If this is 1D or no axis is specified
        return np.prod(arr) ** (1 / arr.size)

multipliers = [1.10, 0.88, 1.90, 0.70, 1.25]
print(f"Geometric Mean: {geometric_mean(multipliers):.4f}")  # Output: 1.0998
```
*(Tip: If your data is in a Pandas DataFrame, convert it first with `df.to_numpy()` or pass a column directly into this function.)*

![(Geometric Mean)](./supporting_files/image3.jpg)

````python
# --- Demo: Geometric Mean Step-by-Step ---
import numpy as np

def geometric_mean(data, axis=None):
    arr = np.array(data)

    if arr.ndim > 1 and axis is not None:
        n = arr.shape[axis]
        return np.prod(arr, axis=axis) ** (1 / n)
    
    return np.prod(arr) ** (1 / arr.size)


# Yearly returns: +10%, -12%, +90%, -30%, +25%
returns_pct = np.array([10, -12, 90, -30, 25])


# Step 1: Turn percentages into growth multipliers (1 + r/100)
multipliers = (returns_pct / 100 ) + 1
print("Growth multipliers:", multipliers)


# Step 2: Calculate Geometric Mean
geo_mean = geometric_mean(multipliers)
print(f"Geometric Mean Multiplier: {geo_mean:.4f} ({(geo_mean - 1)*100:.2f}% avg yearly return)")


# Step 3: Verify final money after 5 years starting from 1,000
final_money = 1000 * (geo_mean ** len(multipliers))
print(f"Final Portfolio Value:     ${final_money:.2f}")
````

## 3.3 Harmonic Mean

### Concept Summary
The **Harmonic Mean** is used when averaging **rates or speeds** (such as `km/h` or financial price ratios). It works by taking the **reciprocal** (flipping a number upside down into `1 / x`, so `5` becomes `1/5`) of every number, finding their normal average, and flipping the result back.

#### Step-by-Step Example: Why Arithmetic Mean Fails on Car Speeds
Suppose you drive from City A to City B (`60 km` away) at **60 km/h**, and drive straight back the same `60 km` in heavy traffic at **20 km/h**:
- **Wrong (Arithmetic Mean):** $\frac{60 + 20}{2} = 40\text{ km/h}$.
- **Why that is wrong:**
  - Going `60 km` at `60 km/h` takes **1 hour**.
  - Returning `60 km` at `20 km/h` takes **3 hours**.
  - Total distance = `120 km`. Total time = `4 hours`. Your true average speed is:
    $$\frac{120\text{ km}}{4\text{ hours}} = \mathbf{30\text{ km/h}}$$
- **Right (Harmonic Mean):**
  $$H = \frac{2}{\frac{1}{60} + \frac{1}{20}} = \frac{2}{0.0167 + 0.05} = \frac{2}{0.0667} = \mathbf{30\text{ km/h}}$$
  It automatically accounts for the fact that you spent more time driving slowly!

#### Formula
$$H = \frac{n}{\sum_{i=1}^{n} \frac{1}{x_i}} = \frac{n}{\frac{1}{x_1} + \frac{1}{x_2} + \dots + \frac{1}{x_n}}$$

#### Pros & Watch-Outs
- **Ignores huge outliers:** One massive number has very little effect on the harmonic mean.
- **Sensitive to tiny values:** One very small number (like `0.01`) will pull the harmonic mean down a lot.
- **No zeros allowed:** Because the formula divides `1 / x`, having a `0` will cause a division-by-zero error (or return `0`).

---

### How to Code Harmonic Mean in NumPy
Make sure to convert your array to decimals (`dtype=float` or `1.0 / arr`) so Python does not chop off decimals into integers (`1 / 2` becoming `0`):

```python
import numpy as np

def harmonic_mean(data, axis=None):
    arr = np.array(data, dtype=float)  # Convert to float to avoid integer division

    if arr.ndim > 1 and axis is not None:
        n = arr.shape[axis]
        return n / np.sum(1.0 / arr, axis=axis)
    
    else:
        return arr.size / np.sum(1.0 / arr)

speeds = [60, 20]
print("Harmonic Mean Speed:", harmonic_mean(speeds))  # Output: 30.0
```
*(Tip: Python's built-in `statistics.harmonic_mean(speeds)` and `scipy.stats.hmean(speeds)` also calculate this.)*

````python
# --- Demo: Harmonic Mean vs Arithmetic Mean ---
import numpy as np
import statistics

def harmonic_mean(data, axis=None):
    arr = np.array(data, dtype=float)
    if arr.ndim > 1 and axis is not None:
        n = arr.shape[axis]
        return n / np.sum(1.0 / arr, axis=axis)
    return arr.size / np.sum(1.0 / arr)

# 1. Round-trip speed check
trip_speeds = np.array([60, 20])

print("Arithmetic Mean of speeds:", np.mean(trip_speeds))       # Output: 40.0 (Wrong!)
print("Harmonic Mean of speeds:  ", harmonic_mean(trip_speeds))  # Output: 30.0 (Right!)

print("Built-in statistics check:", statistics.harmonic_mean(trip_speeds))


# 2. How it handles a huge outlier (500)
pe_ratios = np.array([12, 15, 14, 13, 500])
print("\nWith huge outlier (500):")
print(f"Arithmetic Mean: {np.mean(pe_ratios):.2f}")       # Pulled up to 110.80
print(f"Harmonic Mean:   {harmonic_mean(pe_ratios):.2f}") # Stays steady at 16.59
````

## 3.4 Weighted Arithmetic Mean

### Concept Summary
Sometimes certain numbers are more important than others and should count more toward the final average:
- **Course Grades:** A Final Exam worth `50%` matters much more than Homework worth `10%`.
- **Time-Based Weights:** Newer data from today matters more than old data from last year.
- **Sample Size Weights:** A survey of `10,000` people is much more trustworthy than a survey of `15` people.
- **Reliability Weights:** Giving more trust to a reliable weather station than a faulty one.

#### Formula
$$\bar{x}_w = \frac{\sum_{i=1}^{n} w_i x_i}{\sum_{i=1}^{n} w_i} = \frac{w_1 x_1 + w_2 x_2 + \dots + w_n x_n}{w_1 + w_2 + \dots + w_n}$$
- $x_i$ = each number
- $w_i$ = the weight (importance) given to that number
*(Note: If all weights are equal, this gives the exact same answer as the normal arithmetic mean.)*

**Step-by-Step Example:**
Suppose your scores are: Homework = `80` (weight `0.20`), Midterm = `90` (weight `0.30`), Final = `70` (weight `0.50`):
1. Multiply each score by its weight and add them up:
   $$(80 \times 0.20) + (90 \times 0.30) + (70 \times 0.50) = 16 + 27 + 35 = 78$$
2. Add up the weights:
   $$0.20 + 0.30 + 0.50 = 1.00$$
3. Divide top by bottom: $\frac{78}{1.00} = \mathbf{78.0}$

---

### How to Code Weighted Mean in NumPy
Multiplying matching elements and adding them together is just a **dot product** (`@` or `np.dot()`), so NumPy solves it in one line:
```python
import numpy as np

def weighted_mean(x, w):
    x_arr = np.array(x, dtype=float)
    w_arr = np.array(w, dtype=float)
    return (w_arr @ x_arr) / w_arr.sum()

grades = [80, 90, 70]
weights = [0.20, 0.30, 0.50]
print("Final Weighted Grade:", weighted_mean(grades, weights))  # Output: 78.0
```
*(Tip: NumPy also has a built-in shortcut `np.average(x, weights=w)` that does the exact same math.)*

````python
# --- Demo: Weighted Arithmetic Mean ---
import numpy as np

def weighted_armean(x, w):
    uppersum = w @ x           # Dot product: w1*x1 + w2*x2 + ...
    return uppersum / w.sum()  # Divide by sum of all weights

# 1. Course grade example
scores = np.array([95, 82, 68])         # Homework, Midterm, Final Exam
weights = np.array([0.15, 0.35, 0.50])  # Weights add up to 1.0

print("Normal Arithmetic Mean:", np.mean(scores))
print("Weighted Course Grade: ", weighted_armean(scores, weights))

# 2. Weights that do NOT add up to 1 (dividing by w.sum() handles any weights)
np.random.seed(0)
w_rand = np.random.uniform(0, 1, 10)
x_rand = np.random.uniform(0, 10, 10)

print(f"\nWeighted Mean (random weights): {weighted_armean(x_rand, w_rand):.6f}")
print(f"Check with np.average():        {np.average(x_rand, weights=w_rand):.6f}")
````
