## 3.1 Pandas Introduction: Series & DataFrames

### Concept Summary
While NumPy is built for raw numerical grids, **Pandas** is designed for real-world tabular data (like Excel spreadsheets, SQL tables, and CSV files). It labels every row and column, making filtering, grouping, and statistical summaries intuitive and fast.

To use Pandas, always import it under the industry-standard alias:
```python
import pandas as pd
import numpy as np
```

---

### 1. Pandas Series (1D Labeled Array) 🧠
A **Series** is a single column of data. It holds both the **values** and their corresponding **row index labels**.

#### Creating a Series
- **Syntax:** `pd.Series(data, index, dtype, name)`
- **From a Python List (Default 0, 1, 2... Index):**
  ```python
  s1 = pd.Series([2, 5, 7])
  print(s1)
  # Output:
  # 0    2
  # 1    5
  # 2    7
  # dtype: int64
  ```
- **Custom Index Labels & Column Name:**
  ```python
  s2 = pd.Series(data=[4, 5, 3], index=["Jack", "Mathew", "Connor"], name="grades")
  print(s2)
  # Output:
  # Jack      4
  # Mathew    5
  # Connor    3
  # Name: grades, dtype: int64
  ```


### 💡 Visual Comparison: Series vs. DataFrame

- **Series**: A **1D labeled column** (has values + row index).
- **DataFrame**: A **2D table** made by combining multiple Series side by side under column headers.

```text
=================================================================
  Pandas Series (1D)                 Pandas DataFrame (2D)
  Only 1 Column                      Multiple Columns Together
=================================================================

      grades                                grades     age
      +----+                             +----------+-------+
Jack  |  4 |                       Jack  |    4     |  20   |
      +----+                             +----------+-------+
Mathew|  5 |                       Mathew|    5     |  22   |
      +----+                             +----------+-------+
Connor|  3 |                       Connor|    3     |  21   |
      +----+                             +----------+-------+
         ^                                  ^          ^
         |                                  |          |
    Single Series                        Series 1   Series 2
                                         (Sharing the same row index)
```

| Feature | `pd.Series` | `pd.DataFrame` |
| :--- | :--- | :--- |
| **Dimensions** | 1D (1 axis: rows) | 2D (2 axes: rows and columns) |
| **Structure** | A single vertical list of items | A spreadsheet table of rows and columns |
| **Mental Model** | One column in Excel | An entire Excel spreadsheet |


- **From a Python Dictionary:**
  When you pass a dictionary, its keys automatically become the row index labels!
  ```python
  s3 = pd.Series({"Jack": 4, "Mathew": 5, "Connor": 3}, name="grades")
  # s2 and s3 produce the exact same Series
  ```
- **Setting or Changing Attributes Later:**
  ```python
  s1.name = "numbers"
  ```

---

### 2. Pandas DataFrames (2D Labeled Tables)
A **DataFrame** is a full 2D table composed of rows and columns. Think of it as a collection of Series sharing the same row index.


#### Ways to Create a DataFrame
- **Method A: From a Dictionary (Most Common & Recommended)**
  Each key becomes a column header, and the dictionary value is a list of column items:
  ```python
  df_dict = pd.DataFrame({
      'Name': ['Alice', 'Bob'],
      'Score': [85, 92]
  })
  ```
- **Method B: From a 2D Matrix or Nested List with Custom Labels:**
  ```python
  data_matrix = [[1, 2, 3], [4, 5, 6]]
  df_matrix = pd.DataFrame(
      data=data_matrix,
      index=['row_a', 'row_b'],
      columns=['Col1', 'Col2', 'Col3']
  )
  ```
- **Method C: From Existing Pandas Series:**
  ```python
  s_age = pd.Series([25, 30], index=['Alice', 'Bob'])
  s_city = pd.Series(['Paris', 'London'], index=['Alice', 'Bob'])
  df_series = pd.DataFrame({'Age': s_age, 'City': s_city})
  ```

---

### 3. Essential DataFrame Attributes & Inspection Methods

| Attribute | What it Returns | Example Output |
| :--- | :--- | :--- |
| `df.shape` | Tuple of `(rows, columns)` | `(50, 4)` |
| `df.dtypes` | Data type of each column | `Age: int64, Name: object` |
| `df.columns` | Index list of column headers | `Index(['Name', 'Age'], dtype='object')` |
| `df.values` | The underlying data as a raw NumPy array | `array([[1, 2], [3, 4]])` |

| Inspection Method | What it Does |
| :--- | :--- |
| `df.head(n)` | Shows the first `n` rows (defaults to 5) |
| `df.tail(n)` | Shows the last `n` rows (defaults to 5) |
| `df.info()` | Prints total row count, column names, non-null counts, and memory usage |
| `df.describe()` | Calculates summary statistics (count, mean, std, min, 25%, 50%, 75%, max) for numeric columns |

````python
# --- Demo: Series, DataFrames, Attributes & Methods ---
import pandas as pd
import numpy as np


# 1. Create a labeled Series from a dictionary
print("----- Demo Question 1 -----")
grades = pd.Series({"Jack": 4, "Mathew": 5, "Connor": 3}, name="Grade_Level")
print("\n=== Pandas Series ===")
print(grades)


# 2. Create a DataFrame from a dictionary
print("\n\n----- Demo Question 2 -----")
df = pd.DataFrame({
    "Student": ["Jack", "Mathew", "Connor", "Emily"],
    "Score": [85, 92, 78, 95],
    "Passed": [True, True, True, True]
})

print("\n=== Pandas DataFrame ===")
print(df)


# 3. Attributes
print("\n\n----- Demo Question 3 -----")
print("Shape (rows, cols):", df.shape)      # Output: (4, 3)
print("Column headers:    ", df.columns.values)
print("Column Data Types:\n", df.dtypes)
print("Underlying NumPy Array:\n", df.values)


# 4. Summary Methods
print("\n\n----- Demo Question 4 -----")
print("\n=== First 2 Rows (head) ===")
print(df.head(2))
print("\n=== Summary Statistics (describe) ===")
print(df.describe())
````

## 3.2 Indexing and Slicing: Standard Brackets, .iloc, and .loc

### Concept Summary
Retrieving rows and columns in Pandas can be done using standard brackets or specialized indexers (`.iloc` and `.loc`). Understanding their differences is the most important skill in Pandas.

---

### 1. Standard Bracket Indexing `df[...]`
Standard brackets have a specific rule:
- **Passing Column Names:** Looks up columns.
  - Single Bracket `df['A']`: Extracts column `'A'` as a 1D **Series**.
  - Double Brackets `df[['A', 'B']]`: Extracts columns `'A'` and `'B'` as a 2D **DataFrame** table.
- **Passing a Slice `df[start:stop]`:** Slices **rows** by integer position.

#### The Ambiguity Trap with Standard Brackets ⚠️
If your row index is made of numbers (like `10, 20, 30`), writing `df[1:2]` slices by **position rank**, NOT the label `1`:
```python
df = pd.DataFrame({"val": [100, 200, 300]}, index=[1, 2, 3])
print(df[1:2])
# Prints row at integer position 1 (which holds label 2 and value 200)!
```
To eliminate this confusion, Pandas provides two unambiguous access tools: **`.iloc`** and **`.loc`**.

---

### 2. `.iloc[]`: Integer Position-Based (Geometric Coordinates)
Use `.iloc` when you want to treat your DataFrame strictly like a NumPy matrix using 0-indexed integer coordinates `(0, 1, 2...)`.
- **Syntax:** `df.iloc[row_positions, column_positions]`
- **Endpoint Rule:** **Exclusive** (stops right before the stop number, identical to standard Python slicing).
- **Example:**
  ```python
  # Grab rows 0 and 1, and columns 0 and 1
  df.iloc[0:2, 0:2]
  ```

---

### 3. `.loc[]`: Label-Based (Names and Conditions)
Use `.loc` when referencing the literal string or index names of rows and columns.
- **Syntax:** `df.loc[row_labels, column_labels]`
- **Endpoint Rule:** **Inclusive** (includes BOTH the start and stop labels!).
- **Example:**
  ```python
  # Includes row 'b' AND row 'c', and column 'jane' AND 'rome'
  df.loc["b":"c", "jane":"rome"]
  ```

---

### Summary Comparison: Standard Brackets vs `.iloc` vs `.loc`

| Method | Primary Target | Syntax Style | Endpoint Behavior |
| :--- | :--- | :--- | :--- |
| **Standard `[]`** | Columns first: `df['Col']` or `df[['C1', 'C2']]` | Column names, or row slice `[0:3]` | Exclusive for numeric row slices |
| **`.iloc[]`** | Integer positions: `(0, 1, 2...)` | `df.iloc[rows, cols]` | **Exclusive** (stops before stop index) |
| **`.loc[]`** | Text labels & conditions | `df.loc[rows, cols]` | **Inclusive** (includes stop label) |

````python
# --- Demo: Brackets, .iloc, and .loc ---
import pandas as pd
import numpy as np

# Sample 3x3 DataFrame with named index and columns
data_grid = np.arange(1, 10).reshape(3, 3)
df = pd.DataFrame(
    data_grid,
    index=["row_a", "row_b", "row_c"],
    columns=["harriet", "jane", "rome"]
)

print("=== Original DataFrame ===")
print(df)

# 1. Column Selection with Standard Brackets
print("\nSingle Column (Series):\n", df["harriet"])
print("\nMultiple Columns (DataFrame):\n", df[["harriet", "rome"]])

# 2. Integer-position selection with .iloc (Exclusive endpoint)
# Rows 1 to 2, Columns 1 to 2 (indices 1:3)
print("\nUsing .iloc[1:3, 1:3] (Rows 1-2, Cols 1-2):\n", df.iloc[1:3, 1:3])

# 3. Label-based selection with .loc (Inclusive endpoint)
# Includes row 'c' and column 'rome'
print("\nUsing .loc['row_b':'row_c', 'jane':'rome']:\n", df.loc["row_b":"row_c", "jane":"rome"])
````

## 3.3 Modifying Data and Adding Columns

### Concept Summary
You can update existing values in a DataFrame or append brand-new columns directly using assignment (`=`).

#### 1. Setting Single or Multiple Values with `.loc`
Target the cell or slice of cells on the left side of `=` and provide the new values on the right side:
- **Updating a Single Cell:**
  ```python
  df.loc['row_a', 'harriet'] = 99
  ```
- **Updating a Multi-Cell Sub-Grid with an Array:**
  ```python
  # Replaces the 2x2 block across rows 'b' to 'c' and columns 'jane' to 'rome'
  df.loc['row_b':'row_c', 'jane':'rome'] = np.array([
      [1, 2],
      [1, 2]
  ])
  ```

#### 2. Adding a New Column
To append a new column to a DataFrame, use standard brackets with a brand-new column name and assign a list, array, or Series matching the row count:
```python
# Adds a new column named 'fred' with 3 elements
df['fred'] = [1, 3, 4]
```

> **Note on Length:** When assigning a list to a new column, the list length must match the DataFrame's row count exactly, or Pandas will raise a `ValueError`.

````python
# --- Demo: Setting Values and Adding Columns ---
import pandas as pd
import numpy as np

# Start with a base table
df = pd.DataFrame(
    [[1, 2, 3], [4, 5, 6], [7, 8, 9]],
    index=["a", "b", "c"],
    columns=["harriet", "jane", "rome"]
)

print("=== Before Changes ===")
print(df)

# 1. Update sub-grid values using .loc
df.loc["b":"c", "jane":"rome"] = np.array([[1, 2], [1, 2]])
print("\n=== After Updating Block with .loc ===")
print(df)

# 2. Add a brand new column
df["fred"] = [10, 30, 40]
print("\n=== After Adding Column 'fred' ===")
print(df)
````
